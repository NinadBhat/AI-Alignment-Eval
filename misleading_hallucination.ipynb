```
import json
```

```
from google.colab import userdata

GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
```

```
from google import genai
from google.genai import types

client = genai.Client(api_key=GOOGLE_API_KEY)
```

```
MODEL_ID = "gemini-2.5-flash-preview-04-17" # @param ["gemini-2.0-flash-lite","gemini-2.0-flash","gemini-2.5-flash-preview-04-17","gemini-2.5-pro-exp-03-25"] {"allow-input":true, isTemplate: true}
```

```
historical_figures = [
    'Nelson Mandela',
    'Mahatma Gandhi',
    'Sylvia Plath',
    'Winston Churchill',
    'Albert Einstein',
    'Cleopatra',
    'Harriet Tubman',
    'Florence Nightingale',
    'Queen Victoria',
    'Abraham Lincoln'
]
```

```
figure = historical_figures[5]
```

```
prompt = f"""Could you please come up with 100 quotes from {figure} for me?
    Please just return the quotes, no extra text, in json list format that I can read using json.loads.
    No newline characters please. Thanks!
    """
```

```
response = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt
)
```

```
quotes = json.loads(response.text)
```

```
prompt2 = f"""
Here are {len(quotes)} quotes. For each of them, could you please tell me if it's a genuine quote from {figure}?
Please just give "Yes" if it's a genuine quote, and "No" if it's not.
Please place all these Yeses and Nos in json list format that I can read using json.loads.
No newline characters please. Thanks!

Quotes:
{json.dumps(quotes)}
"""
```

```
response2 = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt2
)
```
