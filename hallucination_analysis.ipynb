```python
import pickle
from collections import defaultdict
import pandas as pd
```

```python
with open('/content/kabir-1.5b.pkl', 'rb') as f:
    records = pickle.load(f)
```

```python
meta = defaultdict(list)
for idea, responses in records.items():
    for i, response in enumerate(responses):
        if "not" not in response.casefold() and "sorry".casefold() not in response.casefold():
            meta[idea].append(i)
            print(response)
```

```python
df = pd.read_csv("/content/draft_Ideology_Fake_quotes.csv")

df = df.rename(
    {
        "Fake_Quote": "quote",
        "Author": "ideology"
    },axis=1,
)
```

```python
a = []
for idea, ddf in df.groupby(by='ideology'):
    if idea not in meta:
        continue
    a.append(ddf['quote'].iloc[meta[idea]])
```

```python
b = pd.concat(a)
```
